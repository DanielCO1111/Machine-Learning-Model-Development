## Introduction

This project explores the application of classic and modern classification algorithms to image datasets. The analysis focuses on benchmark performance using the MNIST and Fashion-MNIST datasets, comparing a manually implemented Feedforward Neural Network with the K-Nearest Neighbors (KNN) algorithm.

### **1. Core Libraries**

```python
import numpy as np
import matplotlib.pyplot as plt
import torch
from torchvision import datasets, transforms
```

### **2. Dataset Analysis: MNIST and Fashion-MNIST**

#### MNIST Dataset

The MNIST dataset consists of 70,000 grayscale images of handwritten digits (0-9). Each image is size-normalized and centered in a 28x28 pixel frame.

#### Fashion-MNIST Dataset

Fashion-MNIST serves as a more complex alternative to MNIST, featuring 70,000 grayscale images of Zalando's article images across 10 categories (e.g., T-shirts, sneakers, bags).

### **3. Feedforward Neural Network Architecture**

The following architecture was implemented using PyTorch for image classification:

- **Input Layer:** 784 units (flattened 28x28 images)
- **Hidden Layer 1:** 128 units with ReLU activation
- **Hidden Layer 2:** 64 units with ReLU activation
- **Output Layer:** 10 units with Log-Softmax activation

```python
from torch import nn, optim
import torch.nn.functional as F

class NeuralNetwork(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(784, 128)
        self.fc2 = nn.Linear(128, 64)
        self.fc3 = nn.Linear(64, 10)

    def forward(self, x):
        x = x.view(x.shape[0], -1)
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        x = F.log_softmax(self.fc3(x), dim=1)
        return x

model = NeuralNetwork()
print(model)
```

### **4. Training and Validation Analysis**

The model was trained using Stochastic Gradient Descent (SGD) and Negative Log-Likelihood Loss. The training process involves monitoring both training and validation loss to ensure optimal convergence.

### **5. Performance Visualization**

Visualizing the training and validation loss curves provides insights into the model's learning progress and potential overfitting.
